```python
# simple line plot

import matplotlib.pyplot as plt
import numpy as np

x = [1, 2, 3, 4, 5]
y = [ 2, 4, 6, 8, 10]

# plt.figure(figsize=(8,6))
plt.plot(x,y)
plt.xlabel(('x Axis'))
plt.ylabel(('y Axis'))
plt.title('simple line plot')
plt.show()
```

```python
# sine wave plot

import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0,10,100)
y = np.sin(x)

plt.figure(figsize=(8,6))
plt.plot(x,y)
plt.xlabel(('x values'))
plt.ylabel(('sin(x)'))
plt.title('sine wave')
plt.grid(True)
plt.show()
```

```python
# multiple line plot

import matplotlib.pyplot as plt
import numpy as np

x =np.linspace(0,10,100)
y1 = np.sin(x)
y2 = np.cos(x)
y3 = np.sin(x) * np.cos(x)


plt.figure(figsize=(8,6))
plt.plot(x,y1,label='sin(x)')
plt.plot(x,y2,label='cos(x)')
plt.plot(x,y3,label='sin(x) * cos(x)')
plt.xlabel(('x Axis'))
plt.ylabel(('y Axis'))
plt.grid(True)
plt.title('multiple line plot')
plt.show()
```

```python
# scatter plots

import matplotlib.pyplot as plt
import numpy as np


np.random.seed(42)
x = np.random.randn(100)
y = 2 * x + np.random.randn(100) * 0.5


plt.figure(figsize=(8,6))
plt.scatter(x,y, alpha=0.6)
plt.xlabel(('x values'))
plt.ylabel(('y values'))
plt.grid(True, alpha=0.3)
plt.show()

```

```python
# scatter plots with different sizes and colours

import matplotlib.pyplot as plt
import numpy as np


np.random.seed(42)
x = np.random.randn(100)
y =  np.random.randn(100)
colours = np.random.randn(100)
sizes =1000* np.random.randn(100)


plt.figure(figsize=(10,6))
plt.scatter(x,y, c=colours, s=sizes, alpha=0.6, cmap="Set1")
plt.xlabel(('x values'))
plt.ylabel(('y values'))
plt.title("scatter plot with color an size variation")
plt.grid(True, alpha=0.3)
plt.show()

```
