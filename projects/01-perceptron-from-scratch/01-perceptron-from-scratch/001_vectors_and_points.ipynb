 w^Tx=0 

describes vectors orthogonal to w?

That's exactly what's happening.

The set:

x:w^Tx=0

is a subspace whose vectors are perpendicular to w.

When we add b:

w^Tx+b=0 

we shift that boundary away from the origin.

So:
Without bias
w^Tx=0 

gives a hyperplane through the origin.

With bias
w^Tx+b=0 

gives a shifted hyperplane.


```python
import numpy as np


x = np.array([80, 3])

w = np.array([2, 1])

b =-5

score = np.dot(w, x) + b

print(score)
```
