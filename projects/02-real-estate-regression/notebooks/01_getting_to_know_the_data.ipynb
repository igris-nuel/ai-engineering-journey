```python
# Building the Cleaning Pipeline (Method Chaining)
import pandas as pd
# .dropna()


path_1 = "../data/mexico-real-estate-1.csv"
path_2 = "../data/mexico-real-estate-2.csv"
path_3 = "../data/mexico-real-estate-3.csv"

 
# .reset_index restore the index of the data


#  Step 2: Cleaning price_usd with .assign() and lambda

df1 = pd.read_csv(path_1).assign(
    price_usd = lambda x: x["price_usd"]
    .str.replace(",","",regex=False)
    .str.replace("$", "", regex=False)
    .astype(float)
).dropna().reset_index(drop=True)

df2 =(
    pd.read_csv(path_2)
    .assign( price_usd = lambda x: x["price_mxn"].div(19))
    .drop(columns = "price_mxn")
    .dropna().reset_index(drop=True)
) 
# print(df2)
coord = pd.read_csv(path_3)["lat-lon"].str.split(",",expand=True).astype(float)

df3 = (
    pd.read_csv(path_3)
    .assign(
        lat = coord[0],
        lon = coord[1],
        state=lambda x: x["place_with_parent_names"].str.split("|",expand=True)[2]
    )
    .dropna()
    .drop(columns=["lat-lon","place_with_parent_names"])
    .reset_index(drop=True)
)

# print(df3)

# columns_match = set(df1) == set(df2) == set(df3)

# if columns_match:
#     print("✅ All DataFrames have the same columns.")
# else:
#     print("❌ Column mismatch detected!")

# print(df1.columns.tolist())
# print(df2.columns.tolist())
# print(df3.columns.tolist())

# get the combination of the 3 csv files
df = pd.concat([df1, df2, df3],ignore_index=True)
# df.info()

df.describe()

# print(df3_1)
```
